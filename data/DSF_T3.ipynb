## Computational thinking

Listing distinct values in the list in order of their first appearance



```python
def find_distinct_values(L: list) -> list:
  output = []
  for element in L:
    if element not in output:
      output.append(element)
  return output
```

```python
my_list = [2, 4, 3, 2, 3]
find_distinct_values(my_list)
```

Abstraction: the proposed algorithm works beyond numbers.

```python
my_list = ['cat', 'dog', 'chicken', 'dog', 'cat']
find_distinct_values(my_list)
```

We'll 'pretend' that our telephone's data is contained under the folder `sample_data`.


The basic steps of data science thinking are: ask, prepare, process, analyse, share, and act.

We'll find the files to delete (**ask**) by collecting statistics about all files in the system (**prepare**), putting them into a table (**process**), and sort them according to their size (**analyse**).

```python
path = './sample_data'

from pathlib import Path

# Collect file names and statistics
file_names = []
file_types = []
file_sizes = []

# Browse all files in the path defined above
for item in Path(path).rglob('*'):
  if item.is_file():
    file_names.append(item.stem)
    file_types.append(item.suffix)
    file_sizes.append(item.stat().st_size/1024**2)  # convert bytes to MB

# Print the three lists
print(file_names)
print(file_types)
print(file_sizes)
```

Great, we now have the required data. Let's put them into a nice table...

```python
import pandas as pd

data = pd.DataFrame(
    {
        'file_name': file_names,
        'file_type': file_types,
        'file_size_in_MB': file_sizes
    }
)

data
```

... and sort the table according to file size in descending order.

```python
data.sort_values(by='file_size_in_MB', ascending=False)
```
